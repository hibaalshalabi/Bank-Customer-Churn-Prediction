```python

# ============================================================
# BANK CUSTOMER CHURN PREDICTION
# MILESTONE 1: EXPLORATORY DATA ANALYSIS (EDA)
# ============================================================

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)

print("=" * 60)
print("MILESTONE 1: EXPLORATORY DATA ANALYSIS")
print("=" * 60)

# ------------------------------------------------------------
# 1. LOAD THE DATASET
# ------------------------------------------------------------

dataset_path = "Churn_Modelling_Uncleaned.csv"

if not os.path.exists(dataset_path):
    try:
        from google.colab import files
        print("Please upload Churn_Modelling_Uncleaned.csv")
        uploaded = files.upload()

        if dataset_path not in uploaded:
            raise FileNotFoundError(
                "Please upload the file named Churn_Modelling_Uncleaned.csv"
            )
    except ImportError:
        raise FileNotFoundError(
            "Place Churn_Modelling_Uncleaned.csv in the same folder as this notebook."
        )

df = pd.read_csv(dataset_path)

print("\n1. DATASET LOADED SUCCESSFULLY")
print("Dataset shape:", df.shape)

# ------------------------------------------------------------
# 2. PREVIEW THE DATA
# ------------------------------------------------------------

print("\n2. FIRST FIVE ROWS")
display(df.head())

print("\n3. LAST FIVE ROWS")
display(df.tail())

# ------------------------------------------------------------
# 3. DATASET STRUCTURE
# ------------------------------------------------------------

print("\n4. DATASET INFORMATION")
df.info()

print("\n5. COLUMN NAMES")
print(df.columns.tolist())

print("\n6. DATA TYPES")
display(df.dtypes.to_frame(name="Data Type"))

# ------------------------------------------------------------
# 4. MISSING VALUES AND DUPLICATES
# ------------------------------------------------------------

print("\n7. MISSING VALUES")
missing_values = df.isnull().sum().to_frame(name="Missing Count")
missing_values["Missing Percentage"] = (
    df.isnull().sum() / len(df) * 100
).round(2)
display(missing_values)

print("\n8. DUPLICATE ROWS")
print("Number of duplicate rows:", df.duplicated().sum())

# ------------------------------------------------------------
# 5. STATISTICAL SUMMARY
# ------------------------------------------------------------

print("\n9. NUMERICAL SUMMARY")
display(df.describe())

print("\n10. CATEGORICAL SUMMARY")
display(df.describe(include="object"))

# ------------------------------------------------------------
# 6. CHURN DISTRIBUTION
# ------------------------------------------------------------

if "Exited" in df.columns:
    print("\n11. CUSTOMER CHURN DISTRIBUTION")

    churn_counts = df["Exited"].value_counts().sort_index()
    churn_percentages = (
        df["Exited"].value_counts(normalize=True).sort_index() * 100
    ).round(2)

    churn_summary = pd.DataFrame({
        "Customer Count": churn_counts,
        "Percentage": churn_percentages
    })

    churn_summary.index = [
        "Stayed" if value == 0 else "Churned"
        for value in churn_summary.index
    ]

    display(churn_summary)

    plt.figure(figsize=(7, 5))
    sns.countplot(data=df, x="Exited")
    plt.title("Customer Churn Distribution")
    plt.xlabel("Exited (0 = Stayed, 1 = Churned)")
    plt.ylabel("Number of Customers")
    plt.tight_layout()
    plt.show()

# ------------------------------------------------------------
# 7. NUMERICAL FEATURE DISTRIBUTIONS
# ------------------------------------------------------------

print("\n12. NUMERICAL FEATURE DISTRIBUTIONS")

numeric_cols = df.select_dtypes(include=np.number).columns.tolist()

# Exclude identifier columns from the distribution plots
id_cols = ["RowNumber", "CustomerId"]
plot_numeric_cols = [
    col for col in numeric_cols if col not in id_cols
]

if plot_numeric_cols:
    df[plot_numeric_cols].hist(
        figsize=(16, 12),
        bins=25,
        edgecolor="black"
    )
    plt.suptitle("Numerical Feature Distributions", fontsize=16)
    plt.tight_layout()
    plt.show()

# ------------------------------------------------------------
# 8. CHURN BY CATEGORICAL FEATURES
# ------------------------------------------------------------

print("\n13. CHURN BY CATEGORICAL FEATURES")

categorical_candidates = ["Geography", "Gender", "NumOfProducts",
                          "HasCrCard", "IsActiveMember"]

categorical_cols = [
    col for col in categorical_candidates if col in df.columns
]

for col in categorical_cols:
    plt.figure(figsize=(7, 5))
    sns.barplot(data=df, x=col, y="Exited", errorbar=None)
    plt.title(f"Churn Rate by {col}")
    plt.xlabel(col)
    plt.ylabel("Churn Rate")
    plt.tight_layout()
    plt.show()

# ------------------------------------------------------------
# 9. AGE AND BALANCE ANALYSIS
# ------------------------------------------------------------

print("\n14. AGE DISTRIBUTION BY CHURN")

if "Age" in df.columns and "Exited" in df.columns:
    plt.figure(figsize=(8, 5))
    sns.boxplot(data=df, x="Exited", y="Age")
    plt.title("Customer Age by Churn Status")
    plt.xlabel("Exited (0 = Stayed, 1 = Churned)")
    plt.ylabel("Age")
    plt.tight_layout()
    plt.show()

print("\n15. BALANCE DISTRIBUTION BY CHURN")

if "Balance" in df.columns and "Exited" in df.columns:
    plt.figure(figsize=(8, 5))
    sns.boxplot(data=df, x="Exited", y="Balance")
    plt.title("Customer Balance by Churn Status")
    plt.xlabel("Exited (0 = Stayed, 1 = Churned)")
    plt.ylabel("Balance")
    plt.tight_layout()
    plt.show()

# ------------------------------------------------------------
# 10. CORRELATION ANALYSIS
# ------------------------------------------------------------

print("\n16. NUMERICAL CORRELATION MATRIX")

if len(numeric_cols) > 1:
    correlation = df[numeric_cols].corr(numeric_only=True)

    plt.figure(figsize=(12, 8))
    sns.heatmap(
        correlation,
        annot=True,
        fmt=".2f",
        cmap="coolwarm",
        center=0
    )
    plt.title("Correlation Matrix")
    plt.tight_layout()
    plt.show()

# ------------------------------------------------------------
# 11. BUSINESS INSIGHTS
# ------------------------------------------------------------

print("\n17. INITIAL BUSINESS INSIGHTS")

if "Exited" in df.columns:
    overall_churn_rate = df["Exited"].mean() * 100
    print(f"Overall churn rate: {overall_churn_rate:.2f}%")

    if "Geography" in df.columns:
        print("\nChurn rate by Geography:")
        print(
            (df.groupby("Geography")["Exited"].mean() * 100)
            .round(2)
            .sort_values(ascending=False)
        )

    if "Gender" in df.columns:
        print("\nChurn rate by Gender:")
        print(
            (df.groupby("Gender")["Exited"].mean() * 100)
            .round(2)
            .sort_values(ascending=False)
        )

    if "IsActiveMember" in df.columns:
        print("\nChurn rate by Activity Status:")
        print(
            (df.groupby("IsActiveMember")["Exited"].mean() * 100)
            .round(2)
        )

    if "Age" in df.columns:
        print("\nAverage age by Churn Status:")
        print(df.groupby("Exited")["Age"].mean().round(2))

print("\nEDA COMPLETED SUCCESSFULLY.")
print("No rows or columns were removed in this notebook.")
print("Data cleaning and preprocessing will be handled separately.")
```
